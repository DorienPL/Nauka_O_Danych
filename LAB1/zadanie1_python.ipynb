```python
# 1. Ładowanie biblioteki Pandas
import pandas as pd
import numpy as np

# Zdefiniowanie nazwy wgranego pliku
file_name = 'IHME_GBD_2019_CHEWING_TOB_1990_2019_DATA_Y2021M05D27.CSV'
```

```python
# 2. Tworzenie ramki danych ze słownika
dict_data = {
    'Imie': ['Anna', 'Jan', 'Piotr'],
    'Wiek': [28, 34, 29],
    'Miasto': ['Warszawa', 'Kraków', 'Poznań']
}
df_dict = pd.DataFrame(dict_data)
print("Ramka danych ze słownika:\n", df_dict, "\n")
```

```python
# 13. Przedstawienie wyboru wierszy i kolumn używając nazw oraz indeksów na różne sposoby
# a) Wybór kolumny przez nazwę
col_val = df['val']
display(col_val)
# b) Wybór wielu kolumn
cols_subset = df[['location_name', 'year_id', 'val']]
display(cols_subset)
# c) Wybór po indeksach liczbowych (iloc) - wiersze od 0 do 4, kolumny od 0 do 2
subset_iloc = df.iloc[0:5, 0:3]
display(subset_iloc)
# d) Wybór po nazwach kolumn i etykietach (loc)
subset_loc = df.loc[0:4, ['location_name', 'val']]
display(subset_loc)
```

```python
# 14. Wybór wierszy pod warunkiem odnośnie określonej wartości kolumny
condition1 = df[df['sex_name'] == 'Male']
display(condition1)
```

```python
# 16. Wybór wierszy, które zawierają w kolumnie kategoryzowanej określone słowo
contains_word = df[df['location_name'].str.contains('Poland', na=False)]
display(contains_word)
```

```python
# 17. Wybór wierszy, które nie zawierają w kolumnie kategoryzowanej określonego słowa
not_contains_word = df[~df['location_name'].str.contains('Global', na=False)]
display(not_contains_word)
```

```python
# 22. Wyświetlenie średniej, maksymalnej i minimalnej wartości z jednej kolumny
mean_val = df['val'].mean()
max_val = df['val'].max()
min_val = df['val'].min()
print(f"Średnia: {mean_val}, Max: {max_val}, Min: {min_val}\n")
```

```python
# 26. Sortowanie wierszy według wartości kolumny (malejąco i rosnąco)
df_sorted_asc = df.sort_values(by='val', ascending=True)
df_sorted_desc = df.sort_values(by='val', ascending=False)
```

```python
# 27. Wyświetlenie 10 największych (najmniejszych) wartości określonej kolumny
top_10 = df.nlargest(10, 'val')
bottom_10 = df.nsmallest(10, 'val')
display(top_10)
display(bottom_10)
```

```python
# 28. Wyświetlenie 10 największych wartości kolumny pod warunkiem określonej wartości innej kolumny
top_10_2019 = df[df['year_id'] == 2019].nlargest(10, 'val')
display(top_10_2019)
```

```python
# 29. Grupowanie według kolumny kategoryzowanej, potem uśrednienie - MultiIndex columns
# Z uwagi na to, że mean() działa tylko na liczbach, wybieramy kolumny numeryczne
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if 'year_id' in numeric_cols: numeric_cols.remove('year_id') # Rok to też liczba, ale chcemy średnią z danych
df_grouped_mean = df.groupby('location_name')[numeric_cols].mean()
display(df_grouped_mean.head())
```

```python
# 30. Grupowanie z wieloma agregacjami na raz (średnia, liczba, mediana)
agg_funcs = {
    'val': 'mean',
    'upper': ['count', 'median']
}
df_grouped_agg = df.groupby('location_name').agg(agg_funcs)
display(df_grouped_agg.head())
```

```python
# 31. Wyświetlenie nazw kolumn indeksu złożonego
print("Nazwy kolumn indeksu złożonego (MultiIndex):")
print(df_grouped_agg.columns)
```

```python
# 32. Sortowanie kolumny w ramce z indeksem złożonym
# Sortujemy po (upper, median)
df_grouped_agg_sorted = df_grouped_agg.sort_values(by=('upper', 'median'), ascending=False)
display(df_grouped_agg_sorted)
```

```python
# 33. Stworzenie tabeli przestawnej (pivot table)
pivot = pd.pivot_table(df, values='val', index=['location_name'], columns=['year_id'], aggfunc='mean')
display(pivot)
```

```python
# 34. Wyświetlenie indeksów i kolumn tabeli przestawnej
print("\nIndeksy tabeli przestawnej:", pivot.index)
print("\nKolumny tabeli przestawnej:", pivot.columns)
```

```python
# 35. Tworzenie indeksu złożonego dla tabeli przestawnej i wyświetlenie go
pivot_multi = pd.pivot_table(df, values='val', index=['location_name', 'sex_name'], columns=['year_id'], aggfunc='mean')
display(pivot_multi.head())
```

```python
# 40. Przedstawienie łączenia ramek danych za pomocą metod merge i concat
df_part1 = df.iloc[:100]
df_part2 = df.iloc[100:200]
# Concat - sklejanie jeden pod drugim
df_concat = pd.concat([df_part1, df_part2])

# Merge - łączenie na podstawie wspólnego klucza (np. location_id)
# Tworzymy dodatkową tabelę (słownik) państw do złączenia
regions = pd.DataFrame({'location_id': [1, 522], 'Region_Type': ['Global', 'Country']})
df_merged = pd.merge(df, regions, on='location_id', how='left')
display(df_merged)
display(df_concat)
```

```python
# 41. Pokazanie dodawania nowych kolumn za pomocą operacji matematycznych
df['rozstep'] = df['upper'] - df['lower']
display(df)
```

```python
# 42. Pokazanie dodawania nowych kolumn za pomocą funkcji lambda
# Przykładowo: Sprawdzenie, czy wartość 'val' jest wyższa niż 0.02
df['is_high'] = df['val'].apply(lambda x: 'Yes' if x > 0.02 else 'No')
display(df)
```

```python
# 43. Przedstawienie możliwości pracy z dużymi plikami przy użyciu argumentu chunksize
chunk_iterator = pd.read_csv(file_name, chunksize=1000, encoding='cp1250')
# Pętla po "kawałkach"
total_rows = 0
for chunk in chunk_iterator:
    # W każdej pętli ramka danych 'chunk' zawiera maksymalnie 1000 wierszy
    total_rows += len(chunk)

print(f"\nPrzetworzono łącznie {total_rows} wierszy przy użyciu chunksize.")
```
